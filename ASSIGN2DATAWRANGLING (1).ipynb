#QUESTION 1

Word embeddings transform words into vector representations that capture the context and semantics of language. These representations allow similar words to appear close to each other in vector space, making them especially powerful in various NLP applications. Word2Vec is a widely-used embedding method that learns these representations by predicting context words from a target word using simple neural network layers.

Common methods include Word2Vec, GloVe, and FastText.

### Key Benefits of Word Embedding:
1. **Semantic Similarity**: Captures the meaning of words based on their usage.
2. **Efficiency**: Reduces dimensionality compared to sparse representations.
3. **Improved NLP Tasks**: Enhances performance in tasks like sentiment analysis and machine translation.

## Step 1: Removing Stop Words

Words such as "I" and "is" are considered stop words — they occur frequently but carry little useful meaning. Removing these words helps improve the quality of word embedding models by focusing learning on semantically meaningful tokens.

## Step 2: Context Window and Training Pairs

We define a context window of size 1, which means for any target word, we consider only one word before and one word after it as context.

## Step 3: Initialising Word Vectors

These vectors are refined during training so that frequently co-occurring words shift closer together in space, while unrelated words move farther apart.

## Step 4: Training – Vector Updates Using Online Learning

The Word2Vec model updates vectors by learning to predict context words given a main word.

For each pair:
1. Compute the dot product between main and context vectors.
2. Apply the sigmoid function to map the score to a probability.
3. Calculate the error:  ERROR=LABEL-SCORE
4. Update both vectors using the rule:  
   $$ V_{\text{new}} = V_{\text{old}} + \eta \cdot \text{ERROR} \cdot V_{\text{other}} $$

Where $$ \eta = 0.1 $$ is the learning rate.

If the error is positive, the vectors are moved closer. If negative, they are pushed apart.

### Example: ("like", "Data")

- Main vector (like) = [-0.4, 0.2]  
- Context vector (Data) = [-0.6, 0.7]  
- Dot product = 0.38  
- Sigmoid(0.38) ≈ 0.594  
- Error = 1 − 0.594 = 0.406

**Vector Updates**:
- New like = [-0.424, 0.228]  
- New Data (context) = [-0.616, 0.708]

## Recurrent Neural Network (RNN)

**Formulas**:
$$
h_t = \sigma(W_h \cdot h_{t-1} + W_x \cdot x_t + b), \quad y_t = W_y \cdot h_t + b
$$

**Assumptions**:
- Identity Fuction: $$ \sigma(x) = x $$
- Learning rate: 0.01
- Initial hidden state: $$ h_{-1} = 0 $$
- Weights: $$ W_h = 0.5, W_x = 1.0, W_y = 1.0, b = 0 $$


### Year 1 (x₀ = 1.1):
$$
h_0 = W_h \cdot h_{-1} + W_x \cdot x_0 + b
$$
$$
h_0 = 0.5 * 0 + 1.0 * 1.1 = 1.1, \quad y_0 = 1.1
$$

### Year 2 (x₁ = 1.2):
$$
h_1 = W_h \cdot h_0 + W_x \cdot x_1 + b
$$
$$
h_1 = 0.5 * 1.1 + 1.0 * 1.2 = 1.75, \quad y_1 = 1.75
$$

### Year 3 (x₂ = 1.0):
$$
h_2 = W_h \cdot h_1 + W_x \cdot x_2 + b
$$
$$
h_2 = 0.5 * 1.75 + 1.0 * 1.0 = 1.875, \quad y_2 = 1.875
$$

### Year 4 Prediction (x₃ = 1.875):
$$
h_3 = W_h \cdot h_2 + W_x \cdot x_3 + b
$$
$$ h_2 = 1.875 $$
$$
h_3 = 0.5 * 1.875 + 1.0 * 1.875 = 2.8125, \quad y_3 = 2.8125
$$

The RNN model uses a simple architecture where each output depends on the current input and the hidden state from the previous time step. The identity activation function is applied, and weights are kept constant (Wh = 0.5, Wx = 1.0, Wy = 1.0), with zero bias and an initial hidden state of 0.

Over three iterations, the hidden state accumulates prior knowledge, and in Year 4 (2025), the model uses the previously predicted value as input to estimate the future price.

The final output from the RNN is 2.8125, indicating a sharp upward trend, as the model strongly reinforces previous values due to its linear accumulation of context without any control mechanism.



### LSTM Model Calculation:
For each time step \( t \):
1. Forget gate:
   $$ f_t = \sigma(W_f \cdot [h_{t-1}, x_t] + b_f) $$
2. Input gate:
   $$ i_t = \sigma(W_i \cdot [h_{t-1}, x_t] + b_i) $$
   $$ \tilde{C}_t = \tanh(W_c \cdot [h_{t-1}, x_t] + b_c) $$
3. Update cell state:
   $$ C_t = f_t \cdot C_{t-1} + i_t \cdot \tilde{C}_t $$
4. Output gate:
   $$ o_t = \sigma(W_o \cdot [h_{t-1}, x_t] + b_o) $$
   $$ h_t = o_t \cdot \tanh(C_t) $$

**Assumptions**:
- All weights = 1, all biases = 0
- Initial state: $$ h_{-1} = 0, C_{-1} = 0 $$
- Activations:
  - $$ \sigma(x) = \frac{1}{1 + e^{-x}} $$
  - $$ \tanh(x) = \frac{e^x - e^{-x}}{e^x + e^{-x}} $$


### Year 1 Calculation:
1. Forget Gate:
   $$ f_1 = \sigma(W_f \cdot h_0 + x_1 + b_f) = \sigma(1 \cdot 0 + 1.1 + 0) = \sigma(1.1) \approx 0.750 $$
2. Input Gate:
   $$ i_1 = \sigma(W_i \cdot h_0 + x_1 + b_i) = \sigma(1 \cdot 0 + 1.1 + 0) = \sigma(1.1) \approx 0.750 $$
3. Candidate Cell State:
   $$ \tilde{C}_1 = \tanh(W_c \cdot h_0 + x_1 + b_c) = \tanh(1 \cdot 0 + 1.1 + 0) = \tanh(1.1) \approx 0.800 $$
4. Cell State:
   $$ C_1 = f_1 \cdot C_0 + i_1 \cdot \tilde{C}_1 = 0.750 \cdot 0 + 0.750 \cdot 0.800 = 0.600 $$
5. Output Gate:
   $$ o_1 = \sigma(W_o \cdot h_0 + x_1 + b_o) = \sigma(1 \cdot 0 + 1.1 + 0) = \sigma(1.1) \approx 0.750 $$
6. Hidden State:
   $$ h_1 = o_1 \cdot \tanh(C_1) = 0.750 \cdot \tanh(0.600) \approx 0.750 \cdot 0.537 = 0.403 $$

### Year 2 Calculation:
1. Forget Gate:
   $$ f_2 = \sigma(W_f \cdot h_1 + x_2 + b_f) = \sigma(1 \cdot 0.403 + 1.2 + 0) = \sigma(1.603) \approx 0.832 $$
2. Input Gate:
   $$ i_2 = \sigma(W_i \cdot h_1 + x_2 + b_i) = \sigma(1 \cdot 0.403 + 1.2 + 0) = \sigma(1.603) \approx 0.832 $$
3. Candidate Cell State:
   $$ \tilde{C}_2 = \tanh(W_c \cdot h_1 + x_2 + b_c) = \tanh(1 \cdot 0.403 + 1.2 + 0) = \tanh(1.603) \approx 0.922 $$
4. Cell State:
   $$ C_2 = f_2 \cdot C_1 + i_2 \cdot \tilde{C}_2 = 0.832 \cdot 0.600 + 0.832 \cdot 0.922 = 0.499 + 0.767 = 1.266 $$
5. Output Gate:
   $$ o_2 = \sigma(W_o \cdot h_1 + x_2 + b_o) = \sigma(1 \cdot 0.403 + 1.2 + 0) = \sigma(1.603) \approx 0.832 $$
6. Hidden State:
   $$ h_2 = o_2 \cdot \tanh(C_2) = 0.832 \cdot \tanh(1.266) \approx 0.832 \cdot 0.853 = 0.710 $$

### Year 3 Calculation:
1. Forget Gate:
   $$ f_3 = \sigma(W_f \cdot h_2 + x_3 + b_f) = \sigma(1 \cdot 0.710 + 1.0 + 0) = \sigma(1.710) \approx 0.846 $$
2. Input Gate:
   $$ i_3 = \sigma(W_i \cdot h_2 + x_3 + b_i) = \sigma(1 \cdot 0.710 + 1.0 + 0) = \sigma(1.710) \approx 0.846 $$
3. Candidate Cell State:
   $$ \tilde{C}_3 = \tanh(W_c \cdot h_2 + x_3 + b_c) = \tanh(1 \cdot 0.710 + 1.0 + 0) = \tanh(1.710) \approx 0.937 $$
4. Cell State:
   $$ C_3 = f_3 \cdot C_2 + i_3 \cdot \tilde{C}_3 = 0.846 \cdot 1.266 + 0.846 \cdot 0.937 = 1.071 + 0.792 = 1.863 $$
5. Output Gate:
   $$ o_3 = \sigma(W_o \cdot h_2 + x_3 + b_o) = \sigma(1 \cdot 0.710 + 1.0 + 0) = \sigma(1.710) \approx 0.846 $$
6. Hidden State:
   $$ h_3 = o_3 \cdot \tanh(C_3) = 0.846 \cdot \tanh(1.863) \approx 0.846 \cdot 0.952 = 0.806 $$

### Prediction for Year 4 (2025):

1. Forget Gate:
   $$ f_4 = \sigma(W_f \cdot h_3 + x_4 + b_f) = \sigma(1 \cdot 0.806 + 1.875 + 0) = \sigma(2.681) \approx 0.936 $$
2. Input Gate:
   $$ i_4 = \sigma(W_i \cdot h_3 + x_4 + b_i) = \sigma(1 \cdot 0.806 + 1.875 + 0) = \sigma(2.681) \approx 0.936 $$
3. Candidate Cell State:
   $$ \tilde{C}_4 = \tanh(W_c \cdot h_3 + x_4 + b_c) = \tanh(1 \cdot 0.806 + 1.875 + 0) = \tanh(2.681) \approx 0.990 $$
4. Cell State:
   $$ C_4 = f_4 \cdot C_3 + i_4 \cdot \tilde{C}_4 = 0.936 \cdot 1.863 + 0.936 \cdot 0.990 = 1.745 + 0.926 = 2.671 $$
5. Output Gate:
   $$ o_4 = \sigma(W_o \cdot h_3 + x_4 + b_o) = \sigma(1 \cdot 0.806 + 1.875 + 0) = \sigma(2.681) \approx 0.936 $$
6. Hidden State:
   $$ h_4 = o_4 \cdot \tanh(C_4) = 0.936 \cdot \tanh(2.671) \approx 0.936 \cdot 0.990 = 0.926 $$


The LSTM model enhances the traditional RNN by introducing memory cells and gates—forget, input, output, and cell state updates—which help regulate what information should be retained or discarded. This avoids issues like vanishing gradients and improves long-term dependency learning.

All weights and biases were set to 1 and 0 respectively, and sigmoid and tanh functions were used for gate activations and state updates. The initial hidden and cell states were both zero. Through controlled gating, the model refines the influence of past values on each step’s prediction.

The final predicted house price for 2025 using the LSTM model is 0.926, which is significantly lower than the RNN output. This shows that LSTM moderated the influence of increasing trends by balancing both past values and the current input more effectively.

### Conclusion:

The RNN model aggressively amplifies past trends due to its lack of gating, which may lead to overestimation in time-series forecasts. In contrast, the LSTM model provides a more conservative and arguably more realistic forecast by controlling the flow of information through gates, preventing runaway accumulation of hidden state values.

While RNN is simpler and faster to implement, the LSTM model is more reliable for long-term predictions involving time series. It accounts for both recent and distant trends, making it better suited for financial forecasting tasks like house price prediction.

```python
import pandas as pd
import string
import nltk
nltk.download('stopwords')
from nltk.corpus import stopwords

# Load dataset
data = pd.read_csv("Toxic.csv")

# Binarize labels: 1 if any toxicity, else 0
data['label'] = (data[['toxic', 'severe_toxic', 'obscene', 'threat', 'insult', 'identity_hate']]
                 .sum(axis=1) > 0).astype(int)

# Clean text
stop_words = set(stopwords.words('english'))
def clean_text(text):
    text = text.lower().translate(str.maketrans('', '', string.punctuation))
    tokens = text.split()
    return ' '.join([word for word in tokens if word not in stop_words])

data['cleaned'] = data['comment_text'].astype(str).apply(clean_text)
data[['comment_text', 'cleaned', 'label']].head()
```

**Overview:**
The original dataset contains 6 binary labels representing different forms of toxicity. To simplify the task into a binary classification problem, a new label is created where a comment is marked **1 (toxic)** if it contains any of the six labels, and **0 (non-toxic)** otherwise.

**Text cleaning steps include:**
- Converting text to lowercase
- Removing punctuation
- Tokenizing with .split()
- Removing stop words using NLTK's corpus

These cleaned texts are stored in a new column cleaned.

## (b) Word2Vec Embedding and FNN Training

```python
!pip install --upgrade --force-reinstall numpy gensim
```

```python
from gensim.models import Word2Vec
import numpy as np

# Train Word2Vec
sentences = [text.split() for text in data['cleaned']]
word2vec_model = Word2Vec(sentences, vector_size=100, window=5, min_count=5, workers=4)

# Create embedding for each comment
def get_embedding(comment):
    words = comment.split()
    valid_words = [word for word in words if word in word2vec_model.wv]
    if valid_words:
        return np.mean([word2vec_model.wv[word] for word in valid_words], axis=0)
    return np.zeros(100)

data['embedding'] = data['cleaned'].apply(get_embedding)
X = np.vstack(data['embedding'].values)
y = data['label'].values

# FNN
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
# Train FNN with Word2Vec embeddings
fnn = Sequential([
    Dense(64, activation='relu', input_shape=(100,)),
    Dense(64, activation='relu'),
    Dense(1, activation='sigmoid')
])
fnn.compile(optimizer=Adam(0.001), loss='binary_crossentropy', metrics=['accuracy'])
fnn.fit(X, y, epochs=5, batch_size=32)
```

To transform the cleaned comments into numeric representations, a Word2Vec model was trained on the text. Parameters such as vector_size=100, window=5, and min_count=5 were used to control embedding quality and context span. Each comment was then represented by averaging the embeddings of all valid words.

Using these 100-dimensional averaged vectors, a Feedforward Neural Network (FNN) was constructed. The architecture consisted of two dense layers with 64 neurons each, activated by ReLU, and a final sigmoid-activated output layer for binary classification. The model was compiled using binary cross-entropy as the loss function and the Adam optimizer with a learning rate of 0.001. It was trained for five epochs with a batch size of 32.

```python
# Attention Mechanism
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input

inputs = Input(shape=(100,1))
attention = Attention()([inputs, inputs])
flat = Flatten()(attention)
dense = Dense(64, activation='relu')(flat)
output = Dense(1, activation='sigmoid')(dense)
attention_model = Model(inputs=inputs, outputs=output)
attention_model.compile(optimizer=Adam(0.001), loss='binary_crossentropy', metrics=['accuracy'])
attention_model.fit(X_train.reshape(-1, 100, 1), y_train, epochs=5, batch_size=32)
y_pred_am = (attention_model.predict(X_test.reshape(-1, 100, 1)) > 0.5).astype(int)
```

To assess the performance of various machine learning and deep learning models on toxic comment classification, five different models were trained using Word2Vec-embedded input features: Logistic Regression (LR), Feedforward Neural Network (FNN), Recurrent Neural Network (RNN), Long Short-Term Memory (LSTM), and a model with an Attention Mechanism.

Each of these models was built with consistency in training parameters to ensure a fair comparison. The settings included two hidden layers (where applicable), each with 64 neurons and ReLU activation. All models used a single sigmoid neuron in the output layer to perform binary classification. The loss function was binary cross-entropy, the optimizer was Adam with a learning rate of 0.001, and training was conducted for 5 epochs with a batch size of 32. The training and test sets were kept constant across all models with an 80-20 stratified split.

###Feedforward Neural Network (FNN)


The FNN was trained on static 100-dimensional Word2Vec embeddings for each comment. It showed a stable learning curve across all epochs, starting with an accuracy of 93.80% and reaching 95.17% by the fifth epoch. The steady decline in loss from 0.1678 to 0.1293 indicates that the network effectively optimized the weights to separate toxic from non-toxic comments. As expected, FNN demonstrated robust performance due to its ability to capture complex patterns through multiple dense layers, even though it lacks any sequential processing capabilities.

Interpretation: The FNN is well-suited for high-dimensional feature vectors like Word2Vec averages. Despite not modeling word order, its non-linearity allowed it to distinguish between comment types effectively.

###Recurrent Neural Network (RNN)


The RNN model was built using TensorFlow’s SimpleRNN layer to process the embedded data as sequences. It initially achieved 92.22% accuracy but did not show consistent improvement. By the fifth epoch, the accuracy slightly declined to 92.36%, and loss values fluctuated — increasing again in the final epoch to 0.2063. This suggests the RNN suffered from the vanishing gradient problem, which is common in basic RNN architectures when learning long-term dependencies.

Interpretation: RNNs are designed to model temporal dependencies, but the simple RNN structure could not retain long-term information effectively, likely leading to underfitting on complex patterns and reduced generalization.

###Long Short-Term Memory (LSTM)


The LSTM model, an advanced version of RNN, was introduced to overcome the limitations of vanishing gradients by using memory gates. It began training with 91.49% accuracy and steadily improved across epochs, reaching 93.75% by the fifth epoch. The loss reduced from 0.2375 to 0.1710, showing the model’s growing ability to discriminate toxic and non-toxic inputs over time.

Interpretation: LSTM's gated architecture effectively captured long-term dependencies in the sequence data. Its higher training time reflects increased computational complexity, but it delivered better learning and generalization than the simple RNN.

###Attention Mechanism


The attention-based model used TensorFlow’s Attention layer to selectively focus on important parts of the input sequences. This model demonstrated strong and consistent improvement during training, starting at 92.33% and climbing to 94.57% accuracy in just 5 epochs. The loss dropped from 0.2184 to 0.1477, reflecting efficient learning. Compared to the RNN and LSTM, the attention model learned faster and reached a better balance between training speed and accuracy.

Interpretation: Attention helped the model dynamically emphasize relevant tokens in each comment, making it particularly effective at handling variable-length and noisy text data. The model’s efficiency and strong performance make it a competitive choice for toxic comment classification.

###Conclusion


All models were successfully trained using the same embedding vectors and configurations. The FNN and Attention Mechanism emerged as the most effective models in terms of both speed and performance. The FNN benefited from non-linear transformations over dense embeddings, while Attention excelled at focusing on semantically meaningful parts of the text. LSTM followed closely behind, offering a balance between temporal context learning and prediction accuracy. The RNN model underperformed due to its architectural limitations. Logistic Regression, not shown in this log but used earlier, serves as a reliable baseline but lacks the capability to model complex non-linear interactions.

After training all five models using the same input features and hyperparameters, their effectiveness was assessed using four key evaluation metrics: accuracy, precision, recall, and F1-score. These metrics provide a comprehensive understanding of how well each model distinguishes between toxic and non-toxic comments.

###Logistic Regression


The logistic regression model, although the simplest of all, delivered a solid performance with an accuracy of 93.86%. However, its recall was significantly lower at 0.5288, which means it failed to identify nearly half of the actual toxic comments. Consequently, the F1-score — which balances precision and recall — stood at 0.6365. This result confirms that while logistic regression is efficient and interpretable, it struggles to detect all toxic cases, which could be critical in real-world moderation tasks.

###Feedforward Neural Network (FNN)

The FNN model showed the strongest overall performance, achieving the highest accuracy at 94.80% and the best F1-score at 0.7106. With a precision of 0.8193 and recall of 0.6274, the model demonstrated a strong ability to minimize both false positives and false negatives. The improvement over logistic regression is attributed to the FNN’s capability to capture complex, non-linear relationships within the dense Word2Vec embeddings. Its consistent learning curve and robust evaluation metrics suggest that FNN is highly suitable for toxic comment classification, especially when both interpretability and predictive power are needed.

###Recurrent Neural Network (RNN)

The basic RNN model showed moderate performance, with an accuracy of 92.55%, precision of 0.6816, and recall of 0.5008. Its F1-score was the lowest among all models at 0.5774, reflecting an imbalance between its ability to identify toxic comments and its reliability in doing so without excessive false alarms. The sequential nature of RNNs was expected to offer advantages in handling text data; however, the simple RNN architecture likely struggled with long sequences and lost important contextual cues, resulting in reduced recall.

###Long Short-Term Memory (LSTM)

The LSTM model performed better than the simple RNN, with an accuracy of 93.64%, precision of 0.7927, and recall of 0.5066. The attention-based model achieved one of the best performances, second only to FNN, with an accuracy of 94.48% and an F1-score of 0.6826.
